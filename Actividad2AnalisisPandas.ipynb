2. Determina el tipo de datos que tienen las columnas.
- Cambia la columna `medicion` a datetime utilizando el formato: `%d/%m/%y %H:%M`
- Obtén dos columnas adicionales separando `medicion` en `fecha` y `hora`

```python
print("\n------------------------------------------------------------------------")

print(f"Tipo de dato actual de la columna 'medicion': '{df['medicion'].dtypes}'")

df["medicion"] = pd.to_datetime(df["medicion"], format="%d/%m/%y %H:%M")

print(
    f"Después del cambio de formato, la columna 'medicion' tiene un tipo:\
    '{df['medicion'].dtypes}'"
)
```

```python
print("\n------------------------------------------------------------------------")

print(f"Creación de la columna 'fecha'")

df["fecha"] = df["medicion"].dt.date
df["fecha"].head()
```

```python
print("\n------------------------------------------------------------------------")

print(f"Creación de la columna 'hora'")

df["hora"] = df["medicion"].dt.time
df["hora"].head()
```

3. Determina la cantidad de valores únicos de cada columna.
- La columna `medicion` tiene un valor menos que el número total de filas del dataframe, lo que indica que hay una fecha duplicada. Identifica cuál es.
- Comprueba si las demás columnas también contienen los mismos valores; si es así, elimina  uno de los registros duplicados.


```python
print("\n------------------------------------------------------------------------")

print(
    "Revisar ambos registros duplicados (sólo en 'medicion') y "
    "comparar sus otras columnas para saber si hay duplicados:"
)
regDupEnMedicion = df[df.duplicated(subset=["medicion"], keep=False)]

valorRegDupEnMedicion = regDupEnMedicion.duplicated().sum()
print(f"\tExisten {valorRegDupEnMedicion} registro(s) duplicado(s)\n")

if valorRegDupEnMedicion > 0:
    print("Tamaño actual del dataframe:")
    print(f"{df.shape[0]} filas, {df.shape[1]} columnas\n")

    print(
        f"\t\tComo hay {valorRegDupEnMedicion} registro(s) duplicado(s), "
        "lo(s) eliminaremo(s) dejando sólamente el primero"
    )

    indiceRegABorrar = df[df["medicion"].duplicated()]

    print(
        f"\t\tSe elimnará(n) el(los) siguiente(s) registro(s): {list(indiceRegABorrar.index)}"
    )

    df = df.drop_duplicates(subset=["medicion"], keep="first")

print("\nNuevo tamaño del dataframe:")
print(f"{df.shape[0]} filas, {df.shape[1]} columnas")
```
